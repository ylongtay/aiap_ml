```python
import pandas as pd

#Load the dataset 
df = pd.read_csv('../data/resale_transactions.csv')

def fill_missing_names(df: pd.DataFrame, id_column: str,name_column: str) -> pd.DataFrame:
    """
    Fills missing values in the 'name_column' using the 'id_column'.
    Args:
    df (pd.DataFrame): The DataFrame containing the columns to be filled.
        id_column (str): The name of the column containing the IDs.
        name_column (str): The name of the column containing the names to be filled.
    Returns:
        pd.DataFrame: The DataFrame with missing values in 'name_column' filled.
    """
    # Identify missing values in the 'name_column'
    missing_names = df[name_column].isna()

    #missing_ids = df.loc[missing_names, id_column]
    #print(f"IDs for rows with missing '{name_column}':")
    #print(missing_ids.to_list())

    #print(f"Found {missing_names.sum()} missing values in '{name_column}' to fill using '{id_column}'.")
    #print(f"Missing IDs: {df.loc[missing_names, id_column].unique()}")


    # Create a dictionary mapping 'id_column' to 'name_column'
    name_mapping = (df[[id_column,name_column]]
        .dropna()
        .drop_duplicates()
        .set_index(id_column)[name_column]
        .to_dict()
    )

    #print(f"Mapping of '{id_column}' to '{name_column}': {name_mapping}")


    # Fill missing 'name_column' using the mapping
    # For each row with a missing name, it looks up that row’s ID in the dictionary and fills in the corresponding name.
    df.loc[missing_names, name_column] = (
        df.loc[missing_names,id_column].map(name_mapping)
    ) 

    print(f"Filled {missing_names.sum()} missing values in '{name_column}' using '{id_column}'.")
    return df

#Fill missing town names using the town_id
df = fill_missing_names(df, 'town_id', 'town_name')
print(df[['town_id', 'town_name']].head(10))

#Fill missing flat model names using the flat_model_id
df = fill_missing_names(df, 'flatm_id', 'flatm_name')
print(df[['flatm_id', 'flatm_name']].head(10))
```
