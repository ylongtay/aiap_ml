```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

#Load the dataset 
df = pd.read_csv('../data/resale_transactions.csv')
df.drop_duplicates(inplace=True)  # Remove duplicate rows

#Set the aesthetic style of the plots
sns.set_style("whitegrid")

#Bar chart for flat type with annotations
plt.figure(figsize=(10, 6))
count_plot = sns.countplot(x='flat_type', data=df)
plt.title('Count of Different Flat Types', fontsize=16)
plt.xlabel('Flat Type', fontsize=14)
plt.ylabel('Count', fontsize=14)
plt.xticks(rotation=45)

#Annotate the bars with frequency counts
for p in count_plot.patches:
    count_plot.annotate(format(p.get_height(), '.0f'),
                        (p.get_x() + p.get_width() / 2., p.get_height()),
                        ha='center', va='center',
                        xytext=(0, 9), textcoords='offset points')

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

#Load the dataset 
df = pd.read_csv('../data/resale_transactions.csv')
df.drop_duplicates(inplace=True)  # Remove duplicate rows
print(df.describe().T)  # Display summary statistics of the dataset

# Replace 'FOUR ROOM' with '4 ROOM' in the 'flat_type' column
df['flat_type'] = df['flat_type'].replace('FOUR ROOM', '4 ROOM')

#Set the aesthetic style of the plots
sns.set_style("whitegrid")

#Count mounthly resale listings
month_counts = df['month'].value_counts().sort_index()

#Line Chart to visualize the number of resale listings by month
plt.figure(figsize=(14, 7))
line_plot = sns.lineplot(x=month_counts.index, y=month_counts.values,marker = 'o')

#Annotate the Data Points 
for x, y in zip(month_counts.index, month_counts.values):
    plt.text(x, y, f'{y}', color='green', ha='center', va='bottom')

#Label the Chart 
plt.title('Trend of Resale Listings by Month', fontsize=16)
plt.xlabel('Month', fontsize=14)
plt.ylabel('Number of Resale Listings', fontsize=14)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
